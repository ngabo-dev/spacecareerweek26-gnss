## Stage 2: Load GP JSON
In this stage, we will define a function to load the General Perturbations (GP) JSON data. GP data provides orbital parameters (SGP4 mean elements) that we will use to compute the orbits. We will save the data file inside `IN_DIR` and load its contents.

```python
def load_gp_data(filepath_or_url):
    """
    Loads GP dataset from a local path or downlods it from a URL,
    saves it to IN_DIR, and returns the parsed JSON list.
    """
    path = Path(filepath_or_url)
    local_file = IN_DIR / path.name

    # If it looks like a URL, fetch it
    if str(filepath_or_url).startswith(('http://', 'https://')):
        response = requests.get(filepath_or_url)
        response.raise_for_status()
        data = response.json()
        with open(local_file, 'w', encoding='utf-8') as f:
            json.dump(data, f, indent=2)
        print(f"Downloaded and saved: {local_file.name}")
    else:
        # Load from local file
        if not path.exists():
            raise FileNotFoundError(f"Local file not found: {path}")
        with open(path, 'r', encoding='utf-8') as f:
            data = json.load(f)
        # Copy to IN_DIR if not already there
        if path.parent != IN_DIR:
            with open(local_file, 'w', encoding='utf-8') as f:
                json.dump(data, f, indent=2)
            print(f"Copied file to input directory: {local_file.name}")
        else:
            print(f"Loaded existing file: {local_file.name}")

    print(f"Successfully loaded {len(data)} GP record(s).")
    return data

# Placeholder demonstration of the loader with mock GP data to ensure it runs correctly
mock_gp_data = [
    {
        "OBJECT_NAME": "ISS (ZARYA)",
        "OBJECT_ID": "1998-067A",
        "EPOCH": "2026-03-20T12:00:00.000000",
        "MEAN_MOTION": 15.49547848,
        "ECCENTRICITY": 0.0004312,
        "INCLINATION": 51.6412,
        "RA_OF_ASC_NODE": 120.3411,
        "ARG_OF_PERICENTER": 45.1234,
        "MEAN_ANOMALY": 315.8765
    }
]

mock_file_path = IN_DIR / "mock_gp.json"
with open(mock_file_path, "w", encoding="utf-8") as f:
    json.dump(mock_gp_data, f, indent=2)

gp_records = load_gp_data(mock_file_path)
print("First GP Record Key Info:", {k: gp_records[0][k] for k in ["OBJECT_NAME", "EPOCH"] if k in gp_records[0]})
```

```python
import os

def handle_live_group():
    """
    Checks if LIVE_GROUP is set. If so, downloads live OMM JSON data from CelesTrak
    under the specified group name, caches it locally for 2 hours, and updates
    the active 'sats' and 'source_note' global variables with the first four satellites.
    """
    global sats, source_note

    if not LIVE_GROUP:
        print("LIVE_GROUP is set to None. Using the offline workshop snapshot data.")
        return

    cache_file = IN_DIR / f"gp_{LIVE_GROUP.lower()}.json"
    cache_duration_hours = 2
    should_download = True

    if cache_file.exists():
        file_age_seconds = time.time() - cache_file.stat().st_mtime
        if file_age_seconds < cache_duration_hours * 3600:
            should_download = False
            print(f"Using cached live dataset from {cache_file.name} (last modified {file_age_seconds/60:.1f} minutes ago).")

    if should_download:
        url = f"https://celestrak.org/NORAD/elements/gp.php?GROUP={LIVE_GROUP}&FORMAT=JSON"
        print(f"Fetching live group '{LIVE_GROUP}' from CelesTrak...")
        try:
            response = requests.get(url, timeout=30)
            response.raise_for_status()
            live_records = response.json()

            if not isinstance(live_records, list) or len(live_records) == 0:
                raise ValueError("Received empty or invalid JSON response from CelesTrak.")

            with open(cache_file, 'w', encoding='utf-8') as f:
                json.dump(live_records, f, indent=2)
            print(f"Successfully downloaded and cached live group: {cache_file.name}")
        except Exception as e:
            print(f"Error downloading from CelesTrak: {e}")
            if cache_file.exists():
                print("Falling back to existing expired cache.")
            else:
                print("Falling back to default workshop snapshot datasets.")
                return

    # Read from cache
    with open(cache_file, 'r', encoding='utf-8') as f:
        group_records = json.load(f)

    # Validate and sort by NORAD catalog ID
    valid_records = []
    for r in group_records:
        if 'NORAD_CAT_ID' in r and 'OBJECT_NAME' in r:
            valid_records.append(r)

    valid_records.sort(key=lambda x: int(x['NORAD_CAT_ID']))

    if len(valid_records) < 1:
        print("No valid records found in the downloaded group. Keeping snapshot data.")
        return

    # Take the first four records
    selected_records = valid_records[:4]
    sats = selected_records

    # Update global source and manifest note
    retrieval_time = datetime.fromtimestamp(cache_file.stat().st_mtime, tz=timezone.utc).strftime('%Y-%m-%dT%H:%M:%SZ')
    source_note = f"Source: CelesTrak Live Group: '{LIVE_GROUP}' (Cached: {retrieval_time})"

    print(f"\nActivated {len(sats)} satellite(s) from live group '{LIVE_GROUP}':")
    for sat in sats:
        print(f"- {sat['OBJECT_NAME']} (NORAD ID: {sat['NORAD_CAT_ID']}, Epoch: {sat['EPOCH']})")

handle_live_group()
```

### Why OMM JSON is easier than the TLE format
OMM JSON is significantly easier to work with than the traditional line-based TLE format because it uses clearly named fields and structured ISO 8601 datetimes instead of strict character-column position mapping. Additionally, it natively supports six-digit catalog numbers without the legacy formatting workarounds required by standard TLE files.

```python
def extract_inputs():
    """
    Checks if the required input files already exist in IN_DIR.
    If not, polls /content/ for 'orbit_workshop_inputs*.zip', waits until fully uploaded,
    and extracts exactly 'orbital_elements.json', 'data_manifest.json', and 'world_map.txt' into IN_DIR.
    """
    required_files = ['orbital_elements.json', 'data_manifest.json', 'world_map.txt']
    all_exist = all((IN_DIR / f).exists() for f in required_files)

    if all_exist:
        print("All required input files already exist in the input directory.")
        return

    # If not all exist, prompt user and wait for zip upload
    print("--- Action Required ---")
    print("Please drag 'orbit_workshop_inputs.zip' into the Colab Files panel (into /content).")
    print("Waiting for upload...")

    content_dir = Path('/content')
    zip_path = None
    timeout_seconds = 900  # 15 minutes
    start_time = time.time()

    while True:
        if time.time() - start_time > timeout_seconds:
            raise TimeoutError("Upload timeout. Please upload 'orbit_workshop_inputs.zip' to /content and try again.")

        # Search for any zip matching orbit_workshop_inputs*.zip
        zips = list(content_dir.glob('orbit_workshop_inputs*.zip'))
        if zips:
            potential_zip = zips[0]
            # Ensure file is completely written and valid
            if zipfile.is_zipfile(potential_zip):
                zip_path = potential_zip
                break

        time.sleep(2.0)

    print(f"Found valid zip archive: {zip_path.name}. Extracting required inputs...")

    with zipfile.ZipFile(zip_path, 'r') as zip_ref:
        for file_info in zip_ref.infolist():
            name = Path(file_info.filename).name
            if name in required_files:
                target_path = IN_DIR / name
                with zip_ref.open(file_info) as source, open(target_path, 'wb') as target:
                    target.write(source.read())
                print(f"Extracted: {name} -> {target_path}")

    # Verify extraction
    missing = [f for f in required_files if not (IN_DIR / f).exists()]
    if missing:
        raise FileNotFoundError(f"Extraction failed. Missing files: {missing}")
    print("Input files extraction completed successfully.")


def read_gp_json(path):
    """
    Loads a CelesTrak OMM JSON list and validates that each record contains
    all required Keplerian mean elements and metadata.
    """
    if not path.exists():
        raise FileNotFoundError(f"Required file not found: {path}")

    with open(path, 'r', encoding='utf-8') as f:
        records = json.load(f)

    required_keys = [
        'OBJECT_NAME', 'NORAD_CAT_ID', 'EPOCH', 'MEAN_MOTION',
        'ECCENTRICITY', 'INCLINATION', 'RA_OF_ASC_NODE',
        'ARG_OF_PERICENTER', 'MEAN_ANOMALY'
    ]

    for i, record in enumerate(records):
        missing = [key for key in required_keys if key not in record]
        if missing:
            raise ValueError(f"Record at index {i} is missing required keys: {missing}")

    return records

# Execute Stage 2 pipeline
extract_inputs()

# Load Keplerian elements
sats = read_gp_json(IN_DIR / 'orbital_elements.json')

# Load and display Manifest details
manifest_path = IN_DIR / 'data_manifest.json'
with open(manifest_path, 'r', encoding='utf-8') as f:
    data_manifest = json.load(f)

source_note = f"Source: {data_manifest.get('source', 'CelesTrak')} (Retrieved: {data_manifest.get('retrieved_utc', 'Unknown UTC')})"
print(f"\n{source_note}")
print(f"Total loaded satellites: {len(sats)}")
print("\nSatellite Details:")
for sat in sats:
    print(f"- {sat['OBJECT_NAME']} (NORAD ID: {sat['NORAD_CAT_ID']}, Epoch: {sat['EPOCH']})")
```

## Stage 3: Parse Elements

We convert raw CelesTrak GP JSON parameters to our model units (km, rad, UTC-aware datetime).

### GP Field to Keplerian Element Mapping
| CelesTrak Field | Keplerian Symbol | Model Unit | Conversion Method |
| :--- | :---: | :---: | :--- |
| `EPOCH` | $t_0$ | UTC Datetime | ISO 8601 parsing with UTC timezone |
| `MEAN_MOTION` | $n$ | rad/s | $\text{MEAN_MOTION} \times 2\pi / 86400$ |
| — | $a$ | km | Kepler's Third Law: $a = (\mu_{earth} / n^2)^{1/3}$ |
| `ECCENTRICITY` | $e$ | Dimensionless | Direct decimal conversion |
| `INCLINATION` | $i$ | rad | Degrees to Radians |
| `RA_OF_ASC_NODE`| $\Omega$ | rad | Degrees to Radians |
| `ARG_OF_PERICENTER`| $\omega$ | rad | Degrees to Radians |
| `MEAN_ANOMALY` | $M_0$ | rad | Degrees to Radians |

*Note: GP values are SGP4 mean elements containing simplified perturbation modeling, so our two-body propagation is an educational approximation rather than a high-precision trajectory prediction.*

```python
def parse_gp_epoch(text):
    """
    Parses a CelesTrak ISO 8601 epoch string into a UTC-aware datetime.
    """
    return datetime.fromisoformat(text.rstrip('Z')).replace(tzinfo=timezone.utc)


def parse_gp_elements(rec):
    """
    Converts a raw General Perturbations (GP) JSON record into Keplerian
    elements in model units (km, rad/s, radians, UTC-aware datetime).
    """
    name = rec['OBJECT_NAME']
    norad_id = int(rec['NORAD_CAT_ID'])
    label = f"{name} [{norad_id}]"
    epoch = parse_gp_epoch(rec['EPOCH'])

    # Convert mean motion from revolutions/day to radians/second
    n_rad_s = (2.0 * math.pi * float(rec['MEAN_MOTION'])) / 86400.0

    # Kepler's Third Law to find semi-major axis (a_km)
    a_km = (MU_EARTH_KM3_S2 / (n_rad_s ** 2)) ** (1.0 / 3.0)

    # Parse remaining Keplerian elements
    e = float(rec['ECCENTRICITY'])
    i_rad = math.radians(float(rec['INCLINATION']))
    raan_rad = math.radians(float(rec['RA_OF_ASC_NODE']))
    argp_rad = math.radians(float(rec['ARG_OF_PERICENTER']))
    M0_rad = math.radians(float(rec['MEAN_ANOMALY']))

    return {
        'name': name,
        'norad_id': norad_id,
        'label': label,
        'epoch': epoch,
        'n_rad_s': n_rad_s,
        'a_km': a_km,
        'e': e,
        'i_rad': i_rad,
        'raan_rad': raan_rad,
        'argp_rad': argp_rad,
        'M0_rad': M0_rad
    }


def elements_table(elements_list):
    """
    Generates and returns lines of a text table summarizing the parsed orbital elements.
    """
    header = f"{'Satellite Name':<28} | {'NORAD':<5} | {'Epoch (UTC)':<26} | {'a [km]':<10} | {'e':<9} | {'i [°]':<7} | {'RAAN [°]':<8} | {'ArgP [°]':<8} | {'M0 [°]':<8}"
    separator = "-" * len(header)
    lines = [header, separator]

    for el in elements_list:
        line = (f"{el['name']:<28} | "
                f"{el['norad_id']:<5} | "
                f"{str(el['epoch']):<26} | "
                f"{el['a_km']:<10.3f} | "
                f"{el['e']:<9.7f} | "
                f"{math.degrees(el['i_rad']):<7.3f} | "
                f"{math.degrees(el['raan_rad']):<8.3f} | "
                f"{math.degrees(el['argp_rad']):<8.3f} | "
                f"{math.degrees(el['M0_rad']):<8.3f}")
        lines.append(line)

    return lines

# Run the stage 3 logic and save results to 'elements'
if 'sats' not in globals() or not sats:
    raise NameError("Missing 'sats' variable from Stage 2. Please execute Stage 2 first.")

elements = [parse_gp_elements(s) for s in sats]

# Print the summary table
table_lines = elements_table(elements)
for line in table_lines:
    print(line)
```

## Stage 4: Time Window

To perform our orbital comparisons on a uniform timeline, we construct a single global UTC time grid shared by all satellites.

### Why Time Alignment and Epoch Offsets Matter
Each satellite's Keplerian orbital elements are specified at a specific historical moment (its state epoch). Since we use an approximate analytical model (two-body Keplerian), the simulation error accumulates over time relative to that epoch. Simulating a satellite far from its initial state epoch results in degraded accuracy. For instance, in our dataset, the **QZS-2** record dates from October 17, 2026—several weeks prior to the other satellites' epochs in November 2026—making its propagation approximation significantly less precise.

```python
# Ensure elements from Stage 3 exist
if 'elements' not in globals() or not elements:
    raise NameError("Missing 'elements' list from Stage 3. Please run Stage 3 first.")

# 1. Determine simulation start: newest epoch in elements, floored to UTC midnight
newest_epoch = max(el['epoch'] for el in elements)
start = newest_epoch.replace(hour=0, minute=0, second=0, microsecond=0)

# 2. Determine simulation end
end = start + timedelta(hours=DURATION_HOURS)

# 3. Generate simulation grid of UTC datetimes
total_seconds = int((end - start).total_seconds())
n_points = total_seconds // STEP_SECONDS + 1

times = [start + timedelta(seconds=i * STEP_SECONDS) for i in range(n_points)]

# 4. Print time window summary
print("=== Stage 4: Time Grid Constructed successfully ===")
print(f"Simulation Start (UTC): {start}")
print(f"Simulation End (UTC):   {end}")
print(f"Total Grid Points:      {n_points:,}\n")

# 5. Check epoch offset for each satellite and warn if stale
print("Satellite Epoch Offsets relative to Simulation Start:")
for el in elements:
    offset_td = start - el['epoch']
    offset_days = offset_td.total_seconds() / 86400.0
    stale_flag = "[STALE RECORD WARNING]" if abs(offset_days) > 7.0 else "[OK]"
    print(f"- {el['name']:<28} | Epoch: {str(el['epoch'])} | Offset: {offset_days:6.2f} days | {stale_flag}")
```

## Stage 5: Propagate

In this stage, we implement a simple analytical two-body Keplerian propagator to simulate our satellites. For each time step:
1. Solve Kepler's Equation for the Eccentric Anomaly ($E$) from the Mean Anomaly ($M$) using Newton-Raphson iteration:
   $$E - e \sin E = M$$
2. Compute the position coordinates in the satellite's orbital plane.
3. Rotate the coordinates into the Earth-Centered Inertial (ECI) frame using the inclination ($i$), RAAN ($\Omega$), and Argument of Pericenter ($\omega$).
4. Perform a Greenwich Meridian rotation to transform ECI positions into the Earth-Centered Earth-Fixed (ECEF) rotating frame to obtain realistic ground track points.

## Stage 6: Orbit Summary

In this stage, we calculate the key orbital characteristics of our propagated satellites. By analyzing their semi-major axes, eccentricities, inclinations, and simulated speed profiles, we can categorize their orbit classifications (such as Low Earth Orbit, Medium Earth Orbit, Geostationary, or Inclined Geosynchronous Orbit).

```python
def orbit_summary(el, pts):
    """
    Computes a physical summary of the satellite's orbital parameters and speed ranges.
    """
    a = el['a_km']
    e = el.get('e', el.get('ecc'))
    i_deg = math.degrees(el.get('i_rad', el.get('inc_rad')))
    n_rad_s = el.get('n_rad_s', el.get('mean_motion_rad_s'))

    # Period in hours: T = 2 * pi / n
    period_h = (2.0 * math.pi / n_rad_s) / 3600.0

    # Perigee & Apogee altitudes (km above spherical Earth)
    perigee_km = a * (1.0 - e) - EARTH_RADIUS_KM
    apogee_km = a * (1.0 + e) - EARTH_RADIUS_KM

    # Extract speeds from simulated trajectory points
    speeds = [pt['speed_km_s'] for pt in pts]
    min_speed = min(speeds) if speeds else 0.0
    max_speed = max(speeds) if speeds else 0.0

    return {
        'name': el['name'],
        'norad_id': el['norad_id'],
        'period_h': period_h,
        'perigee_km': perigee_km,
        'apogee_km': apogee_km,
        'e': e,
        'i_deg': i_deg,
        'min_speed_km_s': min_speed,
        'max_speed_km_s': max_speed
    }

def summary_table(summaries_list):
    """
    Generates and returns lines of a text summary table of all orbital characteristics.
    """
    header = f"{'Satellite Name':<25} | {'NORAD':<5} | {'Period [h]':<10} | {'Perigee [km]':<12} | {'Apogee [km]':<11} | {'e':<9} | {'i [°]':<7} | {'V_min [km/s]':<12} | {'V_max [km/s]'}"
    separator = "-" * len(header)
    lines = [header, separator]

    for s in summaries_list:
        line = (f"{s['name']:<25} | "
                f"{s['norad_id']:<5} | "
                f"{s['period_h']:<10.2f} | "
                f"{s['perigee_km']:<12.1f} | "
                f"{s['apogee_km']:<11.1f} | "
                f"{s['e']:<9.7f} | "
                f"{s['i_deg']:<7.2f} | "
                f"{s['min_speed_km_s']:<12.3f} | "
                f"{s['max_speed_km_s']:.3f}")
        lines.append(line)
    return lines

# Generate orbit summaries
summaries = [orbit_summary(el, tracks[el['norad_id']]) for el in elements]

# Validate Galileo 29 period assert if present
galileo_summary = next((s for s in summaries if s['norad_id'] == 59598), None)
if galileo_summary:
    assert 14.0 <= galileo_summary['period_h'] <= 14.2, f"Galileo 29 period {galileo_summary['period_h']:.2f}h is outside [14.0, 14.2]!"

# Print summary table
for line in summary_table(summaries):
    print(line)
```

## Stage 7: Plots

In this stage, we visualize our propagated satellite trajectories using `matplotlib`. We will generate two key plots:
1. **Ground Tracks**: A 2D flat-map projection overlaying the satellites' longitude and latitude coordinates on top of a gray Earth coastline background.
2. **Velocity Time Series**: A stacked timeline of orbital speeds showing how orbital height, eccentricity, and energy conservation (the Vis-Viva equation) govern satellite velocities over time.

```python
def load_coastline(path):
    """
    Loads coastline segments from world_map.txt.
    The file contains 'latitude longitude' pairs (latitude first) per line,
    with blank lines separating individual continuous coastline segments.
    Returns a list of segments, where each segment is a list of (lon, lat) tuples.
    """
    if not path.exists():
        raise FileNotFoundError(f"Coastline file not found: {path}")

    segments = []
    current_segment = []

    with open(path, 'r', encoding='utf-8') as f:
        for line in f:
            line = line.strip()
            if not line:
                if current_segment:
                    segments.append(current_segment)
                    current_segment = []
            else:
                parts = line.split()
                if len(parts) >= 2:
                    lat = float(parts[0])
                    lon = float(parts[1])
                    current_segment.append((lon, lat))

    if current_segment:
        segments.append(current_segment)

    print(f"Successfully loaded {len(segments)} coastline segment(s) from {path.name}.")
    return segments

# Define Stage 7 global colors (consistent with Stage 8)
COLORS = ['tab:blue', 'tab:orange', 'tab:green', 'tab:red']
```

### Physical Interpretation of Ground Tracks and Velocity Profiles

#### 1. Ground Track Behavior
* **PULSAR-0**: Shows **many diagonal passes shifting west**. This is a typical Low Earth Orbit (LEO) pattern. Because the satellite's orbital period is very short (~1.58 hours) compared to the Earth's 24-hour rotation, the Earth rotates eastward underneath the satellite, causing successive ground tracks to shift westward over time.
* **GSAT0225 (Galileo 29)**: Traces out a **wide, sweeping sinusoidal wave**. As a Medium Earth Orbit (MEO) navigation satellite with a ~14.08-hour period, it completes nearly two full orbits in a day, tracing a highly regular wave covering intermediate-to-high latitudes.
* **SES-5**: Appears as a **single, stationary dot** on the map. Because its orbital period matches the Earth's rotational period exactly (geosynchronous) and it has a near-zero inclination and circular orbit, it stays fixed over its designated longitude (~5.0° East).
* **QZS-2**: Traces an **asymmetric figure-8** ground track. Because it is geosynchronous, its longitude remains bounded. However, its $39^\circ$ inclination causes it to move north and south, while its eccentric orbit ($e \approx 0.075$) creates an asymmetric shape. Because of Kepler's laws, it spends the majority of its time lingering slowly near its northern apogee (benefiting users in Japan) and sweeps quickly through its southern perigee.

#### 2. Velocity Variations and the Vis-Viva Equation
* According to the **Vis-Viva equation** ($v = \sqrt{\mu(2/r - 1/a)}$), a satellite's speed depends entirely on its distance $r$ from Earth's center.
* For circular orbits like **PULSAR-0**, **Galileo 29**, and **SES-5**, $r$ is nearly constant, keeping their speeds virtually flat.
* For the eccentric orbit of **QZS-2**, the speed varies dramatically, peaking at **perigee** (lowest point, highest kinetic energy) and dropping to its minimum at **apogee** (highest point, highest potential energy).

```python
import math
from datetime import datetime, timezone

def solve_kepler(M, e, tol=1e-12, max_iter=50):
    """
    Solves Kepler's Equation: E - e * sin(E) = M for Eccentric Anomaly (E)
    using Newton-Raphson iteration. Raises ValueError if it does not converge.
    """
    # Initial guess
    E = math.pi if e > 0.8 else M
    for _ in range(max_iter):
        f = E - e * math.sin(E) - M
        df = 1.0 - e * math.cos(E)
        delta = f / df
        E -= delta
        if abs(delta) < tol:
            return E
    raise ValueError(f"Kepler solver failed to converge for M={M}, e={e}")

def gmst_rad(t):
    """
    Greenwich Mean Sidereal Time, IAU-1982 (Meeus eq. 12.4).
    Returns angle in radians.
    """
    # J2000 epoch is 2000-01-01 12:00 UTC
    j2000 = datetime(2000, 1, 1, 12, 0, 0, tzinfo=timezone.utc)
    d = (t - j2000).total_seconds() / 86400.0
    T = d / 36525.0

    # IAU-1982 GMST equation in degrees
    theta_deg = (280.46061837 + 360.98564736629 * d + 0.000387933 * T**2 - (T**3) / 38710000.0) % 360.0
    return math.radians(theta_deg)

def propagate(el, sim_times):
    """
    Propagates a satellite's orbit over a list of UTC datetimes.
    """
    trajectory = []
    a = el['a_km']
    e = el.get('e', el.get('ecc')) # Support both key styles from stages
    i = el.get('i_rad', el.get('inc_rad'))
    raan = el['raan_rad']
    argp = el['argp_rad']
    M0 = el.get('M0_rad', el.get('ma_rad'))
    n = el.get('n_rad_s', el.get('mean_motion_rad_s'))
    epoch = el['epoch']

    for t in sim_times:
        # 1) M = M0 + n*(t - epoch), modulo 2*pi
        dt = (t - epoch).total_seconds()
        M = (M0 + n * dt) % (2.0 * math.pi)

        # 2) E = solve_kepler(M, e)
        E = solve_kepler(M, e)

        # 3) r = a*(1 - e*cos E), nu = atan2(sqrt(1-e**2)*sin E, cos E - e), u = argp + nu
        r = a * (1.0 - e * math.cos(E))
        sin_E = math.sin(E)
        cos_E = math.cos(E)
        nu = math.atan2(math.sqrt(1.0 - e**2) * sin_E, cos_E - e)
        u = argp + nu

        # 4) Inertial position coordinates in ECI frame
        cos_u = math.cos(u)
        sin_u = math.sin(u)
        cos_raan = math.cos(raan)
        sin_raan = math.sin(raan)
        cos_i = math.cos(i)
        sin_i = math.sin(i)

        x = r * (cos_raan * cos_u - sin_raan * sin_u * cos_i)
        y = r * (sin_raan * cos_u + cos_raan * sin_u * cos_i)
        z = r * sin_u * sin_i

        # 5) Rotate by theta = gmst_rad(t) into Earth-fixed axes
        theta = gmst_rad(t)
        cos_theta = math.cos(theta)
        sin_theta = math.sin(theta)

        x_fixed = cos_theta * x + sin_theta * y
        y_fixed = -sin_theta * x + cos_theta * y
        z_fixed = z

        # 6) Spherical coordinate conversion
        lat_deg = math.degrees(math.atan2(z_fixed, math.hypot(x_fixed, y_fixed)))
        lon_deg = math.degrees(math.atan2(y_fixed, x_fixed))
        alt_km = r - EARTH_RADIUS_KM

        # 7) Vis-viva speed
        speed_km_s = math.sqrt(MU_EARTH_KM3_S2 * (2.0 / r - 1.0 / a))

        trajectory.append({
            'time': t,
            'lat_deg': lat_deg,
            'lon_deg': lon_deg,
            'alt_km': alt_km,
            'speed_km_s': speed_km_s
        })
    return trajectory

# --- Validation Asserts ---
# Verify GMST at J2000 is approximately 280.46 degrees
j2000_epoch = datetime(2000, 1, 1, 12, 0, 0, tzinfo=timezone.utc)
assert abs(math.degrees(gmst_rad(j2000_epoch)) - 280.46061837) < 1e-5, "GMST math mismatch!"

# Verify Kepler residual (E - e*sin E - M) is near 0
for test_M, test_e in [(0.5, 0.1), (1.5, 0.5), (3.0, 0.05)]:
    res_E = solve_kepler(test_M, test_e)
    assert abs(res_E - test_e * math.sin(res_E) - test_M) < 1e-10, f"Kepler residual too high for e={test_e}!"

# Run propagation for elements on the timeline 'times' (or 'timesteps' depending on cell run)
sim_grid = globals().get('times', globals().get('timesteps', globals().get('sim_times')))
tracks = {el['norad_id']: propagate(el, sim_grid) for el in elements}

# Check geostationary orbit alignment for SES-5 (NORAD ID: 38652)
if 38652 in tracks:
    for pt in tracks[38652]:
        lon = pt['lon_deg']
        assert 4.0 <= lon <= 6.0, f"SES-5 longitude drift detected out of expected limits! Value: {lon:.3f}"

print("=== Stage 5: Propagation Successful and Asserts Passed ===")
for el in elements:
    track = tracks[el['norad_id']]
    print(f"\nSatellite: {el['name']} (NORAD {el['norad_id']})")
    for idx, name in [(0, 'First'), (len(track)//2, 'Middle'), (-1, 'Last')]:
        pt = track[idx]
        print(f"  - {name:<6} Point | UTC: {pt['time']} | Lat: {pt['lat_deg']:7.3f}° | Lon: {pt['lon_deg']:8.3f}° | Alt: {pt['alt_km']:9.2f} km | Speed: {pt['speed_km_s']:.3f} km/s")
```

```python
def solve_kepler(M, e, tol=1e-9, max_iter=100):
    """
    Solves Kepler's Equation: E - e * sin(E) = M for Eccentric Anomaly (E)
    using Newton-Raphson iteration.
    """
    E = M
    for _ in range(max_iter):
        delta = (E - e * math.sin(E) - M) / (1.0 - e * math.cos(E))
        E -= delta
        if abs(delta) < tol:
            break
    return E


def kepler_propagate(el, t_utc, sim_start):
    """
    Propagates a satellite from its state epoch to a target UTC time.
    Returns ECI and ECEF coordinates (in km) and the geodetic latitude/longitude/altitude.
    """
    # Time since state epoch (seconds)
    dt_epoch = (t_utc - el['epoch']).total_seconds()

    # Mean anomaly at target time
    M = (el['M0_rad'] + el['n_rad_s'] * dt_epoch) % (2.0 * math.pi)

    # Solve for Eccentric Anomaly
    E = solve_kepler(M, el['e'])

    # Position in orbital plane
    a = el['a_km']
    e = el['e']
    x_plane = a * (math.cos(E) - e)
    y_plane = a * math.sqrt(1.0 - e**2) * math.sin(E)

    # Direction cosines / rotation matrix components to ECI
    cos_raan = math.cos(el['raan_rad'])
    sin_raan = math.sin(el['raan_rad'])
    cos_argp = math.cos(el['argp_rad'])
    sin_argp = math.sin(el['argp_rad'])
    cos_inc = math.cos(el['i_rad'])
    sin_inc = math.sin(el['i_rad'])

    P_x = cos_raan * cos_argp - sin_raan * sin_argp * cos_inc
    P_y = sin_raan * cos_argp + cos_raan * sin_argp * cos_inc
    P_z = sin_argp * sin_inc

    Q_x = -cos_raan * sin_argp - sin_raan * cos_argp * cos_inc
    Q_y = -sin_raan * sin_argp + cos_raan * cos_argp * cos_inc
    Q_z = cos_argp * sin_inc

    x_eci = x_plane * P_x + y_plane * Q_x
    y_eci = x_plane * P_y + y_plane * Q_y
    z_eci = x_plane * P_z + y_plane * Q_z

    # Greenwich rotation to ECEF
    # Earth rotation speed is ~360.9856 deg/day (7.292115e-5 rad/s)
    OMEGA_EARTH = 7.2921151467e-5
    dt_start = (t_utc - sim_start).total_seconds()
    theta_g = (OMEGA_EARTH * dt_start) % (2.0 * math.pi)

    cos_tg = math.cos(theta_g)
    sin_tg = math.sin(theta_g)

    x_ecef = x_eci * cos_tg + y_eci * sin_tg
    y_ecef = -x_eci * sin_tg + y_eci * cos_tg
    z_ecef = z_eci

    # Spherical Earth coordinate conversion to Geodetic lat, lon (degrees) and alt (km)
    r = math.sqrt(x_ecef**2 + y_ecef**2 + z_ecef**2)
    lat_rad = math.asin(z_ecef / r) if r > 0 else 0.0
    lon_rad = math.atan2(y_ecef, x_ecef)

    lat_deg = math.degrees(lat_rad)
    lon_deg = math.degrees(lon_rad)
    alt_km = r - EARTH_RADIUS_KM

    return {
        'eci': (x_eci, y_eci, z_eci),
        'ecef': (x_ecef, y_ecef, z_ecef),
        'lat': lat_deg,
        'lon': lon_deg,
        'alt': alt_km
    }

# Run the global propagation loop across our time grid for all satellites
results = {}
for el in elements:
    results[el['norad_id']] = {
        'name': el['name'],
        'label': el['label'],
        'trajectory': []
    }
    for t_utc in times:
        state = kepler_propagate(el, t_utc, start)
        results[el['norad_id']]['trajectory'].append(state)

print("=== Stage 5: Propagation Loop Completed successfully ===")
for norad_id, info in results.items():
    final_pos = info['trajectory'][-1]
    print(f"- {info['name']:<28} | Propagated {len(info['trajectory']):,} steps. Final Alt: {final_pos['alt']:.2f} km. Final Lat/Lon: {final_pos['lat']:.4f}°, {final_pos['lon']:.4f}°")
```

## Stage 3: Parse Elements
In this stage, we will define a parsing function to process the raw CelesTrak OMM records into physically meaningful mathematical parameters. Specifically, we will:
1. Parse the ISO 8601 epoch string into a UTC-aware Python `datetime` object.
2. Convert all orbital angles (Inclination, Right Ascension of Ascending Node, Argument of Pericenter, and Mean Anomaly) from degrees to radians.
3. Compute the Semi-Major Axis ($a$ in km) from the Mean Motion ($n$ in revolutions/day) using Kepler's Third Law:

## Stage 3: Parse Elements
In this stage, we will define a parsing function to process the raw CelesTrak OMM records into physically meaningful mathematical parameters. Specifically, we will:
1. Parse the ISO 8601 epoch string into a UTC-aware Python `datetime` object.
2. Convert all orbital angles (Inclination, Right Ascension of Ascending Node, Argument of Pericenter, and Mean Anomaly) from degrees to radians.
3. Compute the Semi-Major Axis ($a$ in km) from the Mean Motion ($n$ in revolutions/day) using Kepler's Third Law:
$$a = \left(\frac{\mu_{earth}}{n^2}\right)^{1/3}$$
where $n$ is first converted to radians/second.

```python
def parse_satellite_elements(sat_record):
    """
    Parses a raw satellite GP JSON record containing CelesTrak OMM field names
    into Keplerian orbital elements using standard physics units (radians, km, UTC datetime).
    """
    # Parse epoch to UTC-aware datetime
    epoch_str = sat_record['EPOCH']
    # Handle common format variations by stripping potential 'Z' and parsing
    clean_epoch_str = epoch_str.replace('Z', '')
    epoch_dt = datetime.fromisoformat(clean_epoch_str).replace(tzinfo=timezone.utc)

    # Parse base identifiers
    name = sat_record['OBJECT_NAME']
    norad_id = int(sat_record['NORAD_CAT_ID'])

    # Convert orbital angles from degrees to radians
    inc_rad = math.radians(float(sat_record['INCLINATION']))
    raan_rad = math.radians(float(sat_record['RA_OF_ASC_NODE']))
    argp_rad = math.radians(float(sat_record['ARG_OF_PERICENTER']))
    ma_rad = math.radians(float(sat_record['MEAN_ANOMALY']))

    # Eccentricity
    ecc = float(sat_record['ECCENTRICITY'])
    if ecc < 0 or ecc >= 1:
        raise ValueError(f"Invalid eccentricity {ecc} for satellite {name}. Only closed elliptical orbits are supported.")

    # Calculate Semi-Major Axis (a) from Mean Motion
    # Mean motion (n_rev_day) is given in revolutions per day.
    n_rev_day = float(sat_record['MEAN_MOTION'])
    # Convert revolutions/day to radians/second
    n_rad_sec = (n_rev_day * 2.0 * math.pi) / 86400.0

    # Kepler's Third Law: a^3 = MU / n^2 => a = (MU / n^2)^(1/3)
    a_km = (MU_EARTH_KM3_S2 / (n_rad_sec ** 2)) ** (1.0 / 3.0)

    # Validate periapsis altitude to make sure the satellite is not inside the Earth
    periapsis_alt = a_km * (1.0 - ecc) - EARTH_RADIUS_KM
    if periapsis_alt < 0:
         print(f"Warning: Satellite {name} has a negative periapsis altitude ({periapsis_alt:.2f} km)!")

    parsed_elements = {
        'name': name,
        'norad_id': norad_id,
        'epoch': epoch_dt,
        'a_km': a_km,
        'ecc': ecc,
        'inc_rad': inc_rad,
        'raan_rad': raan_rad,
        'argp_rad': argp_rad,
        'ma_rad': ma_rad,
        'mean_motion_rad_s': n_rad_sec
    }
    return parsed_elements

# Parse all loaded satellites and print their processed physical elements
parsed_sats = [parse_satellite_elements(sat) for sat in sats]

print("=== Stage 3: Parse Elements Completed Successfully ===")
for psat in parsed_sats:
    print(f"\nSatellite: {psat['name']} (NORAD ID: {psat['norad_id']})")
    print(f"  Epoch (UTC):        {psat['epoch']}")
    print(f"  Semi-major Axis:    {psat['a_km']:.3f} km (Altitude: {psat['a_km'] - EARTH_RADIUS_KM:.2f} km)")
    print(f"  Eccentricity:       {psat['ecc']:.8f}")
    print(f"  Inclination:        {math.degrees(psat['inc_rad']):.4f}° ({psat['inc_rad']:.6f} rad)")
    print(f"  RAAN:               {math.degrees(psat['raan_rad']):.4f}°")
    print(f"  Arg of Pericenter:  {math.degrees(psat['argp_rad']):.4f}°")
    print(f"  Mean Anomaly:       {math.degrees(psat['ma_rad']):.4f}°")
```
